## Análise da Matriz de Correlação

A matriz de correlação (`correlation_matrix`) revela as seguintes relações:

### Correlações com 'Pct_Telehealth':

*   **Maior Correlação Negativa**: 'Year' (-0.435). Isso sugere que, ao longo dos anos, a porcentagem de beneficiários que utilizaram telemedicina diminuiu. Isso é um achado interessante e pode refletir mudanças nas políticas de telemedicina ou na adaptação pós-pandemia.
*   **Maior Correlação Positiva**: 'Total_Bene_Telehealth' (0.109). Embora seja uma correlação fraca, indica que, conforme o número total de beneficiários que utilizam telemedicina aumenta, há uma ligeira tendência de aumento na porcentagem de uso de telemedicina.
*   **Correlações Negativas Muito Fracas**: 'Total_Bene_TH_Elig' (-0.000) e 'Total_PartB_Enrl' (-0.008). Isso sugere que o número total de beneficiários elegíveis ou inscritos no Parte B não tem uma relação linear significativa com a porcentagem de uso de telemedicina.

### Multicollinearidade:

Observamos uma **multicollinearidade extremamente alta** entre as variáveis 'Total_Bene_TH_Elig' e 'Total_PartB_Enrl' (0.994). Isso significa que essas duas variáveis estão fortemente relacionadas e medem aspectos muito semelhantes, o que pode causar problemas em modelos de regressão, pois elas fornecem informações redundantes. Além disso, 'Total_Bene_Telehealth' também apresenta alta correlação com 'Total_Bene_TH_Elig' (0.880) e 'Total_PartB_Enrl' (0.851), indicando que essas três variáveis estão interligadas e capturam a magnitude do universo de beneficiários.

```python
df['High_Adoption'] = df.groupby('quarter')['Pct_Telehealth'].transform(lambda x: x >= x.quantile(0.75)).astype(int)

print("DataFrame com a nova coluna 'High_Adoption':")
print(df.head())
```

```python
X = df.drop('High_Adoption', axis=1)
y = df['High_Adoption']

print("Features (X) head:")
print(X.head())
print("\nTarget variable (y) head:")
print(y.head())
```

```python
categorical_cols = ['quarter', 'Bene_Geo_Desc', 'Bene_Mdcd_Mdcr_Enrl_Stus', 'Bene_Race_Desc', 'Bene_Sex_Desc', 'Bene_Mdcr_Entlmt_Stus', 'Bene_Age_Desc', 'Bene_RUCA_Desc']
numerical_cols = ['Year', 'Total_Bene_TH_Elig', 'Total_PartB_Enrl', 'Total_Bene_Telehealth', 'Pct_Telehealth']

# Ensure all specified columns exist in X
missing_cat_cols = [col for col in categorical_cols if col not in X.columns]
missing_num_cols = [col for col in numerical_cols if col not in X.columns]

if missing_cat_cols:
    print(f"Warning: Categorical columns not found in X: {missing_cat_cols}")
if missing_num_cols:
    print(f"Warning: Numerical columns not found in X: {missing_num_cols}")

print("Categorical columns identified:", categorical_cols)
print("Numerical columns identified:", numerical_cols)
```

```python
X_categorical_processed = pd.get_dummies(X[categorical_cols], drop_first=True)

print("Features categóricas pré-processadas (X_categorical_processed) head:")
print(X_categorical_processed.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_numerical_scaled = scaler.fit_transform(X[numerical_cols])
X_numerical_processed = pd.DataFrame(X_numerical_scaled, columns=numerical_cols, index=X.index)

print("Features numéricas padronizadas (X_numerical_processed) head:")
print(X_numerical_processed.head())
```

```python
X_processed = pd.concat([X_categorical_processed, X_numerical_processed], axis=1)

print("Features pré-processadas combinadas (X_processed) head:")
print(X_processed.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, stratify=y, random_state=42)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_curve, auc, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

print("Required libraries imported successfully.")
```

```python
def evaluate_model(model, X_train, y_train, X_test, y_test, model_name):
    print(f"\n{'='*50}\nEvaluating {model_name} Model\n{'='*50}")

    # a. Fit the model to the training data
    model.fit(X_train, y_train)

    # b. Make predictions (y_pred) on the test data
    y_pred = model.predict(X_test)

    # c. Calculate predicted probabilities (y_proba) for the positive class
    y_proba = model.predict_proba(X_test)[:, 1]

    # d. Print Accuracy, Precision, Recall, F1-score, and Classification Report
    print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precision: {precision_score(y_test, y_pred):.4f}")
    print(f"Recall: {recall_score(y_test, y_pred):.4f}")
    print(f"F1-Score: {f1_score(y_test, y_pred):.4f}")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    # e. Print the Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    print("\nConfusion Matrix:")
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Predicted 0', 'Predicted 1'], yticklabels=['Actual 0', 'Actual 1'])
    plt.title(f'Confusion Matrix for {model_name}')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.show()

    # f. Calculate False Positive Rate (FPR), True Positive Rate (TPR), and thresholds for ROC curve, along with AUC
    fpr, tpr, thresholds = roc_curve(y_test, y_proba)
    roc_auc = auc(fpr, tpr)
    print(f"ROC AUC: {roc_auc:.4f}")

    return fpr, tpr, roc_auc

print("evaluate_model function defined successfully.")
```

```python
logistic_model = LogisticRegression(random_state=42, solver='liblinear') # Using 'liblinear' solver for better convergence with small datasets
fpr_lr, tpr_lr, auc_lr = evaluate_model(logistic_model, X_train, y_train, X_test, y_test, "Logistic Regression")
```

```python
random_forest_model = RandomForestClassifier(random_state=42)
fpr_rf, tpr_rf, auc_rf = evaluate_model(random_forest_model, X_train, y_train, X_test, y_test, "Random Forest Classifier")
```

```python
plt.figure(figsize=(8, 6))
plt.plot(fpr_lr, tpr_lr, color='darkorange', lw=2, label=f'Logistic Regression (AUC = {auc_lr:.4f})')
plt.plot(fpr_rf, tpr_rf, color='green', lw=2, label=f'Random Forest (AUC = {auc_rf:.4f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

```python
feature_importances = random_forest_model.feature_importances_
feature_names = X_processed.columns

importances_df = pd.DataFrame({'feature': feature_names, 'importance': feature_importances})
importances_df = importances_df.sort_values(by='importance', ascending=False)

print("Top 10 Most Important Features from Random Forest:")
print(importances_df.head(10))
```
